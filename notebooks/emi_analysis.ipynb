```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv("../data/emi_dataset.csv")

#veri setinin ilk beş satırı
df.head()
```

```python
#veri setine bakalım
df.info()
df.describe()
```

```python
#kac adet fail kart var
df["Result"].value_counts()
```

```python
#grafiksel dagilimi
df["Result"].value_counts().plot(kind="bar")

plt.title("PASS / FAIL Dağılımı")
plt.xlabel("Sonuç")
plt.ylabel("Sayısı")

plt.show()
```

```python
#EMI skorları
plt.hist(df["EMI_Score"], bins=20)

plt.title("EMI Score Dağılımı")
plt.xlabel("EMI Skoru")
plt.ylabel("Sayısı")

plt.show()
```

```python
#korelasyon analizi
correlation = df.corr(numeric_only=True)

correlation["EMI_Score"].sort_values(ascending=False)
```
